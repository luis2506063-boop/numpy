```python
import matplotlib.pyplot as plt 
img = plt.imread('beagle.jpg')
img.shape

img[10,10]
```

```python
plt.imshow(img)
plt.axis('off')
plt.show()
```

```python
print("Shape de la imagen:", img.shape)
print("Alto:", img.shape[0])
print("Ancho:", img.shape[1])
print("Canales de color:", img.shape[2])
print("Tipo de dato:", img.dtype)
```

```python
pixel = img[10, 10]

print("Pixel en posición [10, 10]:", pixel)
print("Rojo:", pixel[0])
print("Verde:", pixel[1])
print("Azul:", pixel[2])
```

```python
pixel_centro = img[img.shape[0]//2, img.shape[1]//2]

print("Pixel del centro de la imagen:", pixel_centro)
print("Rojo:", pixel_centro[0])
print("Verde:", pixel_centro[1])
print("Azul:", pixel_centro[2])
```

```python
recorte = img[200:700, 300:800]

plt.imshow(recorte)
plt.axis('off')
plt.show()
```
