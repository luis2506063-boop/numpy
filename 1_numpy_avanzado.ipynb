```python
def generar_matriz_numpy(N):
    matriz = np.random.randint(1, 10, size=(N, N))
    return matriz
```

```python
N = 5

matriz_aleatoria = generar_matriz_numpy(N)

print(matriz_aleatoria)
```

```python
def imprimir_matriz(matriz):
    for fila in matriz:
        fila_formateada = ' '.join(f'{elemento:2}' for elemento in fila)
        print(fila_formateada)
```

```python
print("Matriz de 5x5 con NumPy")
m_5 = generar_matriz_numpy(5)
imprimir_matriz(m_5)

print("============")

print("Matriz de 10x10 con NumPy")
m_10 = generar_matriz_numpy(10)
imprimir_matriz(m_10)

print("============")

print("Matriz de 20x20 con NumPy")
m_20 = generar_matriz_numpy(20)
imprimir_matriz(m_20)
```
